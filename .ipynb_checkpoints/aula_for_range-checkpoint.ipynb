```python
import pandas as pd

dados = {
    "nome": ["Ana", "Pedro", "Maria", "Carlos"],
    "idade": [30, 25, 40, 35],
    "salario": [5000, 4500, 7000, 6500]
}

df = pd.DataFrame(dados)

df.to_csv("pessoas.csv", index=False)
```

```python
dados = [
    {"departamento": "A", "salario": 5000},
    {"departamento": "A", "salario": 6000},
    {"departamento": "B", "salario": 4500},
    {"departamento": "B", "salario": 7000}
]

df = pd.DataFrame(dados)

df.groupby("departamento")["salario"].mean()
```
